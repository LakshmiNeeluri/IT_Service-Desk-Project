```python
import pandas as pd
import numpy as np
from datetime import datetime, timedelta
import random

print("Libraries imported successfully!")
```

```python
N = 50000

print("Number of records to generate:", N)
```

```python
priorities = [
    "Low",
    "Medium",
    "High",
    "Critical"
]

categories = [
    "Hardware",
    "Software",
    "Network",
    "Access Management",
    "Email",
    "Security",
    "Database",
    "Application"
]

departments = [
    "IT",
    "Finance",
    "HR",
    "Sales",
    "Marketing",
    "Operations",
    "Customer Support"
]

channels = [
    "Email",
    "Phone",
    "Portal",
    "Chat"
]

statuses = [
    "Resolved",
    "Closed",
    "In Progress",
    "Open",
    "Pending"
]

print("Categories created successfully!")
```

```python
subcategories = {
    "Hardware": [
        "Laptop",
        "Desktop",
        "Printer",
        "Monitor",
        "Keyboard",
        "Mouse"
    ],
    "Software": [
        "Installation",
        "Application Error",
        "License",
        "Update"
    ],
    "Network": [
        "WiFi",
        "VPN",
        "Internet",
        "LAN"
    ],
    "Access Management": [
        "Password Reset",
        "Account Lock",
        "Access Request",
        "Permission Issue"
    ],
    "Email": [
        "Outlook",
        "Mailbox",
        "Spam",
        "Email Access"
    ],
    "Security": [
        "Malware",
        "Phishing",
        "Security Alert",
        "Endpoint Security"
    ],
    "Database": [
        "Connection",
        "Performance",
        "Access",
        "Query Error"
    ],
    "Application": [
        "Application Crash",
        "Login Issue",
        "Performance",
        "Functional Issue"
    ]
}

print("Subcategories created successfully!")
```

```python
agents = [
    "Rahul Sharma",
    "Priya Reddy",
    "Arjun Kumar",
    "Sneha Patel",
    "Vikram Singh",
    "Ananya Rao",
    "Kiran Kumar",
    "Pooja Sharma",
    "Rohit Verma",
    "Neha Reddy",
    "Amit Das",
    "Divya Nair",
    "Sanjay Rao",
    "Meena Patel",
    "Nikhil Kumar"
]

print("Number of agents:", len(agents))
```

```python
np.random.seed(42)
random.seed(42)

ticket_ids = np.arange(1, N + 1)

priority_data = np.random.choice(
    priorities,
    size=N,
    p=[0.30, 0.45, 0.20, 0.05]
)

category_data = np.random.choice(
    categories,
    size=N,
    p=[0.15, 0.20, 0.18, 0.12, 0.10, 0.08, 0.07, 0.10]
)

department_data = np.random.choice(
    departments,
    size=N
)

channel_data = np.random.choice(
    channels,
    size=N,
    p=[0.30, 0.25, 0.30, 0.15]
)

status_data = np.random.choice(
    statuses,
    size=N,
    p=[0.45, 0.20, 0.15, 0.10, 0.10]
)

agent_data = np.random.choice(
    agents,
    size=N
)

print("Basic ticket data generated successfully!")
```

```python
start_date = pd.Timestamp("2024-01-01")
end_date = pd.Timestamp("2025-12-31")

date_range_days = (end_date - start_date).days

created_dates = [
    start_date + pd.Timedelta(
        days=random.randint(0, date_range_days),
        hours=random.randint(0, 23),
        minutes=random.randint(0, 59)
    )
    for _ in range(N)
]

created_dates = pd.to_datetime(created_dates)

print("Created dates generated successfully!")
print("Minimum date:", created_dates.min())
print("Maximum date:", created_dates.max())
```

```python
sla_targets = {
    "Critical": 4,
    "High": 8,
    "Medium": 24,
    "Low": 48
}

sla_target_data = np.array([
    sla_targets[p]
    for p in priority_data
])

print("SLA targets generated successfully!")
```

```python
resolution_time_data = []

for priority in priority_data:
    if priority == "Critical":
        hours = np.random.uniform(1, 8)
    elif priority == "High":
        hours = np.random.uniform(2, 16)
    elif priority == "Medium":
        hours = np.random.uniform(4, 40)
    else:
        hours = np.random.uniform(6, 72)

    resolution_time_data.append(round(hours, 2))

resolution_time_data = np.array(resolution_time_data)

print("Resolution times generated successfully!")
print("Minimum:", resolution_time_data.min(), "hours")
print("Maximum:", resolution_time_data.max(), "hours")
```

```python
resolved_dates = []

for created, status, resolution_hours in zip(
    created_dates,
    status_data,
    resolution_time_data
):
    if status in ["Resolved", "Closed"]:
        resolved_dates.append(
            created + pd.Timedelta(hours=float(resolution_hours))
        )
    else:
        resolved_dates.append(pd.NaT)

resolved_dates = pd.to_datetime(resolved_dates)

print("Resolved dates generated successfully!")
```

```python
sla_status_data = []

for resolution, target, status in zip(
    resolution_time_data,
    sla_target_data,
    status_data
):
    if status not in ["Resolved", "Closed"]:
        sla_status_data.append("Pending")
    elif resolution <= target:
        sla_status_data.append("Met")
    else:
        sla_status_data.append("Breached")

print("SLA status generated successfully!")
```

```python
satisfaction_data = []

for status, sla_status in zip(status_data, sla_status_data):
    if status in ["Open", "Pending", "In Progress"]:
        satisfaction_data.append(np.nan)
    elif sla_status == "Met":
        satisfaction_data.append(
            round(np.random.uniform(3.5, 5.0), 2)
        )
    else:
        satisfaction_data.append(
            round(np.random.uniform(1.0, 3.5), 2)
        )

satisfaction_data = np.array(satisfaction_data)

print("Customer satisfaction scores generated successfully!")
```

```python
df = pd.DataFrame({
    "Ticket_ID": ticket_ids,
    "Created_Date": created_dates,
    "Resolved_Date": resolved_dates,
    "Priority": priority_data,
    "Category": category_data,
    "Subcategory": [
        random.choice(subcategories[category])
        for category in category_data
    ],
    "Department": department_data,
    "Assigned_Agent": agent_data,
    "Channel": channel_data,
    "Status": status_data,
    "SLA_Target_Hours": sla_target_data,
    "Resolution_Time_Hours": resolution_time_data,
    "SLA_Status": sla_status_data,
    "Customer_Satisfaction": satisfaction_data
})

print("DataFrame created successfully!")
print("Rows:", len(df))
print("Columns:", len(df.columns))
```
